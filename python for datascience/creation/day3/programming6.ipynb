```python
class Car:
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
    def display(self):
        print(f"Make:{self.make}\nModel:{self.model}\nyear:{self.year}")

car1=Car("tesla","hy67","2050")
Car.display(car1)
```

```python
class ElectricCar(Car):
    def __init__(self,make,model,year,batteryCapacity):
        super().__init__(make,model,year)
        self.batteryCapacity=batteryCapacity
    def battery_info(self):
        print(f"battery is {self.batteryCapacity}KWH")
ev1=ElectricCar("BYD","dolphin","2025",43)
ev1.display()
ev1.battery_info()
ev1.make="deppal"
ev1.display()
```

```python
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def area(self):
        return(self.length*self.width)
    def perimeter(self):
        return(2*(self.length+self.width))
rec=Rectangle(3,6)
a1=rec.area()
a2=rec.perimeter()
print(a1,a2)
```

```python
class Circle:
    def __init__(self,radius):
        self.__radius=radius
    def set_radius(self,radius):
        self.__radius=radius
    def get_radius(self):
        return self.__radius
    def area(self):
        return 3.1415*(self.__radius)**2
c1=Circle(7)
print(c1.area())
c2=Circle(8)
print(c2.get_radius())
```

```python
class BankAccount:
    def __init__(self,cash):
        self.__cash=cash
    def return_cashamount(self,__cash):
        print(f"your initial amount is :{self.__cash}")
    def deposit(self,__cash):
        self.__cash+=__cash
        print(f"deposited amount:{__cash}")
    def withdraw(self,__cash):
        if self.__cash<__cash:
            print(f" to be withdrawn amount:{__cash}")
            return "Your account has no money"
        self.__cash-=__cash
        print(f"withdrawn amount:{__cash}")


    def checkBalance(self):
        return(self.__cash)
a1=BankAccount(2000)

a1.deposit(200)
result=a1.withdraw(20000)
print(result)
a1.checkBalance()
```
