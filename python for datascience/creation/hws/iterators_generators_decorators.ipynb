# Iterators, Generators, and Decorators Implementation

```python
# 1. Iterator that returns first n even numbers
class EvenIter:
    def __init__(self, n):
        self.n = n
        self.current = 0
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        value = self.current
        self.current += 2
        self.count += 1
        return value
NUM=EvenIter(5)
for N in NUM:
    print(N)
```

```python
# 2. Fibonacci generator up to n terms
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

fib=fibonacci(5)
for n in fib:
    print(n)
```

```python
# 3. Decorator that prints Start and End
def start_end_decorator(func):
    def wrapper(*args, **kwargs):
        print('Start')
        result = func(*args, **kwargs)
        print('End')
        return result
    return wrapper
def hello():
    print("hello")
```

```python
# 4. Generator to read CSV line by line
def read_csv_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip().split(',')
for row in read_csv_line_by_line('reference.txt'):
    print(row)
```

```python
# 5. Reverse range iterator
class ReverseRange:
    def __init__(self, start, end):
        self.current = end - 1
        self.start = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.start:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value
```

```python
# 7. Decorator to measure memory usage
import tracemalloc

def memory_usage(func):
    def wrapper(*args, **kwargs):
        tracemalloc.start()
        result = func(*args, **kwargs)
        current, peak = tracemalloc.get_traced_memory()
        print(f'Peak memory usage: {peak / 10**6} MB')
        tracemalloc.stop()
        return result
    return wrapper
```

```python
# 8. Infinite prime generator
def prime_generator():
    def is_prime(num):
        if num < 2:
            return False
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                return False
        return True

    num = 2
    while True:
        if is_prime(num):
            yield num
        num += 1
```

```python
# 9. Memoization decorator
def memoize(func):
    cache = {}
    def wrapper(*args):
        if args in cache:
            return cache[args]
        result = func(*args)
        cache[args] = result
        return result
    return wrapper
```

```python
# 10. Generator to filter even numbers
def even_filter(numbers):
    for num in numbers:
        if num % 2 == 0:
            yield num
```

```python
# 11. Sensor readings generator
import random

def sensor_stream():
    while True:
        yield random.uniform(20.0, 30.0)
```

```python
# 12. Decorator to log execution time and arguments
import time

def log_execution(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        print(f'Arguments: {args}, {kwargs}')
        result = func(*args, **kwargs)
        end = time.time()
        print(f'Execution time: {end - start} seconds')
        return result
    return wrapper
```

```python
# 13. Iterator for unique items
class UniqueIterator:
    def __init__(self, items):
        self.items = list(dict.fromkeys(items))
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.items):
            raise StopIteration
        value = self.items[self.index]
        self.index += 1
        return value
```

```python
# 14. Decorator to check positive inputs
def positive_inputs(func):
    def wrapper(*args, **kwargs):
        if any(arg <= 0 for arg in args if isinstance(arg, (int, float))):
            raise ValueError('All numeric inputs must be positive')
        return func(*args, **kwargs)
    return wrapper
```

```python
# 15. Infinite random number generator
def infinite_random():
    while True:
        yield random.random()
```
