```python
# Create a Jupyter Notebook (.ipynb) file with solutions

import nbformat
from nbformat.v4 import new_notebook, new_code_cell

notebook = new_notebook()

cells = []

# 1
cells.append(new_code_cell("""
# 1. Iterator that returns the first n even numbers
class EvenNumbersIterator:
    def __init__(self, n):
        self.n = n
        self.current = 0
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        result = self.current
        self.current += 2
        self.count += 1
        return result
"""))

# 2
cells.append(new_code_cell("""
# 2. Generator that yields Fibonacci sequence up to n terms
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
"""))

# 3
cells.append(new_code_cell("""
# 3. Decorator that prints “Start” before and “End” after a function
def start_end_decorator(func):
    def wrapper(*args, **kwargs):
        print("Start")
        result = func(*args, **kwargs)
        print("End")
        return result
    return wrapper
"""))

# 4
cells.append(new_code_cell("""
# 4. Generator to read a CSV file line by line
def read_csv_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip().split(',')
"""))

# 5
cells.append(new_code_cell("""
# 5. Iterator class that iterates over a range in reverse order
class ReverseRange:
    def __init__(self, start, end):
        self.current = end - 1
        self.start = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.start:
            raise StopIteration
        result = self.current
        self.current -= 1
        return result
"""))

# 6
cells.append(new_code_cell("""
# 6. Generator expression to create list of cubes from 1 to 20
cubes = list(x**3 for x in range(1, 21))
"""))

# 7
cells.append(new_code_cell("""
# 7. Decorator to measure memory usage of a function
import tracemalloc

def memory_usage_decorator(func):
    def wrapper(*args, **kwargs):
        tracemalloc.start()
        result = func(*args, **kwargs)
        current, peak = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        print(f"Current memory usage: {current / 10**6}MB; Peak: {peak / 10**6}MB")
        return result
    return wrapper
"""))

# 8
cells.append(new_code_cell("""
# 8. Generator that yields prime numbers indefinitely
def prime_generator():
    num = 2
    while True:
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                break
        else:
            yield num
        num += 1
"""))

# 9
cells.append(new_code_cell("""
# 9. Decorator to cache results of a function (memoization)
def memoize(func):
    cache = {}
    def wrapper(*args):
        if args in cache:
            return cache[args]
        result = func(*args)
        cache[args] = result
        return result
    return wrapper
"""))

# 10
cells.append(new_code_cell("""
# 10. Generator that filters only even numbers from a list
def filter_even(numbers):
    for num in numbers:
        if num % 2 == 0:
            yield num
"""))

# 11
cells.append(new_code_cell("""
# 11. Generator to simulate a stream of sensor readings
import random
import time

def sensor_stream():
    while True:
        yield round(random.uniform(20.0, 30.0), 2)
        time.sleep(1)
"""))

# 12
cells.append(new_code_cell("""
# 12. Decorator to log execution time and input arguments
import time

def log_execution(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Function: {func.__name__}")
        print(f"Arguments: {args}, {kwargs}")
        print(f"Execution time: {end_time - start_time} seconds")
        return result
    return wrapper
"""))

# 13
cells.append(new_code_cell("""
# 13. Iterator that yields only unique items from a list
class UniqueIterator:
    def __init__(self, data):
        self.data = data
        self.index = 0
        self.seen = set()

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.data):
            item = self.data[self.index]
            self.index += 1
            if item not in self.seen:
                self.seen.add(item)
                return item
        raise StopIteration
"""))

# 14
cells.append(new_code_cell("""
# 14. Decorator to check if inputs are positive numbers
def positive_inputs(func):
    def wrapper(*args, **kwargs):
        for arg in args:
            if not isinstance(arg, (int, float)) or arg <= 0:
                raise ValueError("All inputs must be positive numbers")
        return func(*args, **kwargs)
    return wrapper
"""))

# 15
cells.append(new_code_cell("""
# 15. Generator to simulate an infinite sequence of random numbers
import random

def infinite_random_numbers():
    while True:
        yield random.random()
"""))

notebook['cells'] = cells

file_path = "/mnt/data/generators_iterators_decorators.ipynb"
with open(file_path, "w", encoding="utf-8") as f:
    nbformat.write(notebook, f)

file_path
```
