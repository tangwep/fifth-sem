```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.arange(1,5,0.1)
y=np.square(x)
y_cube=np.power(x,3)
plt.plot(x,y,c='b',alpha=0.7,label="label_1")
plt.plot(x,y_cube,c='r',alpha=0.9,label="label_2")
plt.xlabel("time")
plt.ylabel("sales")
plt.title("time vs sales")
plt.legend()
```

```python
categories=['apples','bananas','cherries','dates']
values=[10,15,7,12]
plt.bar(categories,values,width=0.5,edgecolor='black',color=['red','black','purple','orange'])
plt.xlabel('fruit')
plt.ylabel('quantities')
plt.xticks(rotation=360)
plt.title('fruit inventory')
plt.show()
```

```python
categories=['apples','bananas','cherries','dates']
values=[10,15,7,12]
plt.barh(categories,values,edgecolor='black',color=['red','black','purple','orange'])
plt.ylabel('fruit')
plt.xlabel('quantities')
plt.xticks(rotation=360)
plt.title('fruit inventory')
plt.show()
```

```python
data=np.random.normal(loc=50,scale=10,size=10000)
# create histogram
plt.hist(data,bins=20,color='blue',edgecolor='black')
# plt.hist(data,bins=20,color='blue',kde=True,edgecolor='black')
#labels and titles
plt.xlabel('vlues')
plt.ylabel('frequency')
plt.title('histogram of random data')
```

```python
from sklearn.datasets import make_blobs
x,y=make_blobs(
    n_features=2,n_samples=1000,random_state=42
)
plt.scatter(x[:,0],x[:,1],c=y,cmap='viridis',edgecolors='black')
#labels and title
plt.xlabel('feature 1')
plt.ylabel('feature 2')
plt.title('make_blobs visualization')
plt.colorbar(label='cluster')
plt.show()
```

```python
#visualize 3d data
from sklearn.datasets import make_blobs
X_3d, y = make_blobs(n_features=3, n_samples=1000, random_state=42)
ax = plt.figure().add_subplot(111, projection='3d')
ax.scatter(X_3d[:, 0], X_3d[:, 1], X_3d[:, 2], c=y, cmap='viridis')
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.set_zlabel('Feature 3')
plt.show()
```
